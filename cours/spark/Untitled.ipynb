```python
from pyspark import SparkContext, SparkConf

conf = SparkConf().setAppName("Cours de Spark").setMaster("local[*]")
sc = SparkContext(conf=conf)
```

```python
RDD = sc.parallelize(range(50), 3)
print(RDD.glom().collect())
```

```python
def partition_sum(partition):
    return([sum(partition)])
```

```python
print(RDD.mapPartitions(partition_sum).collect())
```

```python
def add(x, y):
    return(x+y)

print(RDD.reduce(add))

## Second method

print(RDD.reduce(lambda x, y: x+y))
```

```python
RDD = sc.parallelize(range(50))

def odd_key(value):
    if value%2 == 0:
        return(0, value)
    else:
        return(1, value)

RDD1 = RDD.map(odd_key)
print(RDD1.reduceByKey(lambda x, y: x+y).collect())
```

```python
RDD = sc.parallelize(range(50))

def odd_key(value):
    if value%2 == 0:
        return(0, value)
    else:
        return(1, value)

RDD1 = RDD.map(odd_key)
print(RDD1.sortByKey().collect())
```

```python
RDD = sc.parallelize(range(50))

RDD_even = RDD.filter(lambda x: x%2 == 0)
RDD_odd = RDD.filter(lambda x: x%2 == 1)

print(RDD_even.count(), RDD_odd.count())
print(RDD_even.reduce(lambda x, y: x+y), RDD_odd.reduce(lambda x, y: x+y))
```

```python
RDD = sc.parallelize(range(50))
RDD1 = RDD.map(lambda x: (x%2, x))
```

```python
def partitioning(key):
    if key==0:
        return(0)
    else:
        return(1)
```

```python
RDD1 = RDD1.partitionBy(2, partitioning)
print(RDD1.glom().collect())
```
